##Load DataLoad and Preprocess the MNIST Dataset

```python
# Import necessary libraries
import numpy as np
import matplotlib.pyplot as plt
import time
from tensorflow.keras.datasets import mnist
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense
from sklearn.metrics import confusion_matrix, classification_report

# Load MNIST dataset
(X_train, y_train), (X_test, y_test) = mnist.load_data()

# Preprocess the data
X_train = X_train.reshape(-1, 28, 28, 1) / 255.0  # Reshape to (28, 28, 1) and normalize
X_test = X_test.reshape(-1, 28, 28, 1) / 255.0    # Reshape to (28, 28, 1) and normalize

# Convert labels to categorical (one-hot encoding)
y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test, 10)
```

##Define CNN Model, Training, and Display Function

```python
# Define a function to create and train a CNN model with custom hyperparameters and display results
def train_and_display_cnn(filters=32, kernel_size=(3,3), pool_size=(2,2), epochs=5, experiment_name="Experiment"):
    # Create CNN model
    model = Sequential([
        Conv2D(filters=filters, kernel_size=kernel_size, activation='relu', input_shape=(28,28,1)),
        MaxPooling2D(pool_size=pool_size),
        Flatten(),
        Dense(128, activation='relu'),
        Dense(10, activation='softmax')  # 10 classes for digits 0-9
    ])

    # Compile the model
    model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

    # Start timer for training
    start_time = time.time()

    # Train the model
    history = model.fit(X_train, y_train, epochs=epochs, validation_data=(X_test, y_test), verbose=0)

    # Calculate training time
    training_time = time.time() - start_time

    # Evaluate the model
    test_loss, test_acc = model.evaluate(X_test, y_test, verbose=0)

    # Generate predictions for confusion matrix
    y_pred = np.argmax(model.predict(X_test), axis=1)
    y_true = np.argmax(y_test, axis=1)

    # Generate confusion matrix
    conf_matrix = confusion_matrix(y_true, y_pred)

    # Generate classification report
    class_report = classification_report(y_true, y_pred)

    # Display results
    print(f"\n--- {experiment_name} ---")
    print(f"Training Time: {training_time:.2f} seconds")
    print(f"Test Accuracy: {test_acc:.4f}")
    print("Confusion Matrix:")
    print(conf_matrix)
    print("\nClassification Report:")
    print(class_report)
```

##Experiment 1: Default Settings (32 filters, 3x3 kernel, 2x2 pool size)

```python
# Run Experiment 1: Default settings (32 filters, 3x3 kernel, 2x2 pool size)
train_and_display_cnn(filters=32, kernel_size=(3,3), pool_size=(2,2), epochs=5, experiment_name="Experiment 1: Default (32 filters, 3x3 kernel, 2x2 pool)")
```

##Experiment 2: Increasing the Number of Filters (64 filters, 3x3 kernel, 2x2 pool size)

```python
# Run Experiment 2: Increasing the number of filters
train_and_display_cnn(filters=64, kernel_size=(3,3), pool_size=(2,2), epochs=5, experiment_name="Experiment 2: 64 filters, 3x3 kernel, 2x2 pool")
```

##Experiment 3: Larger Kernel Size (32 filters, 5x5 kernel, 2x2 pool size)

```python
# Run Experiment 3: Larger kernel size
train_and_display_cnn(filters=32, kernel_size=(5,5), pool_size=(2,2), epochs=5, experiment_name="Experiment 3: 32 filters, 5x5 kernel, 2x2 pool")
```

##Experiment 4: Larger Pool Size (32 filters, 3x3 kernel, 3x3 pool size)

```python
# Run Experiment 4: Larger pool size
train_and_display_cnn(filters=32, kernel_size=(3,3), pool_size=(3,3), epochs=5, experiment_name="Experiment 4: 32 filters, 3x3 kernel, 3x3 pool")
```

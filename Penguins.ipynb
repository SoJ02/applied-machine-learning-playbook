```python
import seaborn as sns
import matplotlib.pyplot as plt

# Plot histograms for all numerical columns
df.hist(bins=20, figsize=(10, 8))
plt.show()

# Interpretation: Histograms help visualize the distribution of numerical data, such as body mass and flipper length, and detect skewness or outliers.
```

```python
# Boxplot for detecting outliers in body mass
sns.boxplot(data=df, x='body_mass_g')
plt.show()

# Interpretation: Boxplots show the spread of the data, including the interquartile range, and help detect outliers.
```

```python
# Bar chart for the species column
sns.countplot(x='species', data=df)
plt.show()

# Interpretation: Bar charts show the frequency of each category in a categorical variable, useful for analyzing class imbalance.
```

```python
# Correlation heatmap for numerical columns
corr_matrix = df.select_dtypes(include=['number']).corr() # Select only numerical columns for correlation calculation
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm')
plt.show()

# Interpretation: The correlation heatmap helps identify relationships between numerical variables, such as bill length and flipper length, and highlights multicollinearity.
```

```python
# Pairplot for exploring relationships between numerical variables
sns.pairplot(df, hue='species')
plt.show()

# Interpretation: Pairplots allow us to examine relationships between multiple numerical variables and detect patterns or clusters based on species.
```

```python
# Heatmap to visualize missing data
sns.heatmap(df.isnull(), cbar=False, cmap='viridis')
plt.show()

# Interpretation: The heatmap helps visualize where and how much data is missing, guiding decisions on how to handle the missing values.
```

```python
from scipy import stats

# Calculate Z-scores for body mass
df['z_score_body_mass'] = stats.zscore(df['body_mass_g'])

# Filter rows with Z-scores > 3 or < -3 (outliers)
outliers = df[df['z_score_body_mass'].abs() > 3]
# Show the outliers
outliers

# Interpretation: Z-scores help detect outliers by identifying values that are more than 3 standard deviations away from the mean.
```

```python
# Using Interquartile Range (IQR) to detect outliers in body mass
Q1 = df['body_mass_g'].quantile(0.25)
Q3 = df['body_mass_g'].quantile(0.75)
IQR = Q3 - Q1


# Filter out rows that have outliers
outliers = df[((df['body_mass_g'] < (Q1 - 1.5 * IQR)) | (df['body_mass_g'] > (Q3 + 1.5 * IQR)))]
# Show the outliers
outliers

# Interpretation: The IQR method identifies outliers as values that fall outside the interquartile range.
```

```python
# Save the cleaned dataset to a new CSV file
df.to_csv('cleaned_penguins.csv', index=False)

# Interpretation: This saves the cleaned dataset, ensuring that the modifications, such as imputed missing values and removed outliers, are preserved.
```
